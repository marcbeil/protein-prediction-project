```python
import numpy as np
from sklearn.preprocessing import OneHotEncoder
import pandas as pd

# List of standard amino acids
amino_acids = [
    'A', 'R', 'N', 'D', 'C', 
    'E', 'Q', 'G', 'H', 'I', 
    'L', 'K', 'M', 'F', 'P', 
    'S', 'T', 'W', 'Y', 'V'
]

# Initialize the sklearn OneHotEncoder
one_hot_encoder = OneHotEncoder(
    categories=[amino_acids],  # Explicit categories in order
    handle_unknown='ignore',   # Ignore unknown amino acids
)

# Fit the encoder
one_hot_encoder.fit(np.array(amino_acids).reshape(-1, 1))

df = pd.read_csv("../../datasets/v5/train_split.csv")
test_seq = list(df["protein_sequence"][0])
test_seq
```
