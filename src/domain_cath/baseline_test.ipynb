```python
import pandas as pd
import numpy as np
from collections import Counter

def cath_baseline_ci(df, cath_col='cath', n_bootstrap=1000, ci=0.95):
    """Compute majority class baseline with bootstrap CI for CATH hierarchy levels"""
    
    def majority_acc(labels):
        return Counter(labels).most_common(1)[0][1] / len(labels)
    
    def bootstrap_ci(labels):
        accs = [majority_acc(np.random.choice(labels, len(labels))) for _ in range(n_bootstrap)]
        alpha = (1 - ci) / 2
        return np.percentile(accs, [alpha*100, (1-alpha)*100])
    
    results = {}
    labels = df[cath_col].astype(str)
    
    for level in range(1, 5):
        level_labels = ['.'.join(label.split('.')[:level]) for label in labels]
        acc = majority_acc(level_labels)
        ci_low, ci_high = bootstrap_ci(level_labels)
        
        results[f'Level_{level}'] = {
            'accuracy': acc,
            'ci_lower': ci_low, 
            'ci_upper': ci_high,
            'n_classes': len(set(level_labels))
        }
    
    return pd.DataFrame(results).T

# Usage:
results = cath_baseline_ci(df)
print(results.round(3))
```
