```python
import pandas as pd

df = pd.read_csv("../datasets/v5/train_split.csv")
```

```python
df[df["protein_id"] == "2zgi"]
```

```python
import py3Dmol 

domain_index = 10
protein_id = df["protein_id"][domain_index]
domain_start = df["domain_start"][domain_index]
domain_end = df["domain_end"][domain_index]


view = py3Dmol.view(query='pdb:' + protein_id) 
view.setStyle({'cartoon':{'color':'white'}}) 
view.addStyle({'resi': f"{domain_start}-{domain_end}"}, {'cartoon': {'color': 'blue'}})

view
```
