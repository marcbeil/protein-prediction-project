```python
import os
import matplotlib.pyplot as plt
import pandas as pd

# Assume 'df' DataFrame is already loaded and contains a 'protein_length' column.
# If you need to load it, uncomment and adjust the line below:
# df = pd.read_csv("path/to/your/subset_protein_mapped_enhanced_limited_len_600.csv")

# 1. Get the value counts of 'protein_length'
# This gives a Series where the index is the length and the values are their counts.
length_counts = df["protein_length"].value_counts()

# 2. Sort the counts by the protein length (index)
# This is crucial for a meaningful cumulative sum graph.
sorted_length_counts = length_counts.sort_index()

# 3. Calculate the cumulative sum of these counts
cumulative_counts = sorted_length_counts.cumsum()

# 4. Create the plot
plt.figure(figsize=(12, 7)) # Adjust figure size for better visualization
plt.plot(cumulative_counts.index, cumulative_counts.values, marker='o', linestyle='-', markersize=4, color='skyblue')
plt.axvline(x=600, color='red', linestyle='--', label='Length = 600')

plt.title("Cumulative Distribution of Protein Lengths", fontsize=16)
plt.xlabel("Protein Length", fontsize=14)
plt.ylabel("Cumulative Number of Proteins", fontsize=14)
plt.grid(True, linestyle='--') # Add a grid
plt.xticks(rotation=45, ha='right') # Rotate x-axis labels if they overlap
plt.tight_layout() # Adjust layout


plt.savefig(os.path.join("../datasets/v5", 'cumulative_distribution.png'))
plt.show()
```
